```python

# DAY 1: INTRODUCTION TO SCIENTIFIC PYTHON

# Import mathematical functions
import math

# Earth's physical properties
earth_mass = 5.972e24       # kg
earth_radius = 6.371e6      # metres

# Calculate volume of Earth
earth_volume = (4 / 3) * math.pi * earth_radius**3

# Calculate mean density
earth_density = earth_mass / earth_volume

# Display results
print("PLANETARY DATA ANALYSIS")
print("-----------------------")

print("Earth Mass:", earth_mass, "kg")
print("Earth Radius:", earth_radius, "m")
print("Earth Volume:", earth_volume, "m^3")
print("Earth Density:", earth_density, "kg/m^3")
```

```python
import numpy as np

# 1. Create an array of UV doses from 0 to 1000 J/m^2 (in steps of 100)
uv_doses = np.arange(0, 1100, 100)

# 2. Define initial parameters
N0 = 1e8  # 10^8 CFU/mL
k = 0.005  # Inactivation constant

# 3. Vectorized calculation (no 'for' loop needed!)
surviving_cells = N0 * np.exp(-k * uv_doses)

# 4. Calculate log10 reduction (standard in microbiology)
log_reduction = np.log10(N0 / surviving_cells)

# Print results side-by-side
for dose, cells, log_red in zip(uv_doses, surviving_cells, log_reduction):
  print(
      f"Dose: {dose:4d} J/m² | Survivors: {cells:.2e} CFU/mL | Log Reduction:"
      f" {log_red:.2f}"
  )
```
